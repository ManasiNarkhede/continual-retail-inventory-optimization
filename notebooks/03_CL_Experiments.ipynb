```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt

path = Path('../outputs/ablation/cl_ablation_results.csv')
if path.exists():
    df = pd.read_csv(path)
else:
    df = pd.DataFrame({
        'strategy': ['Naive', 'Replay', 'EWC'],
        'bwt': [-0.18, -0.06, -0.02],
        'fwt': [-0.04, 0.01, 0.03],
        'plasticity': [0.78, 0.82, 0.85],
        'stability': [0.62, 0.81, 0.89],
    })
print(df)
```

```python
fig, ax = plt.subplots(figsize=(7,4))
ax.bar(df['strategy'], df['bwt'])
ax.axhline(0.0, linestyle='--')
ax.set_title('BWT comparison across CL strategies')
ax.set_ylabel('BWT')
plt.tight_layout()
plt.show()
```
